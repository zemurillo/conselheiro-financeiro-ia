```python
class DadosPessoais():

    def __init__(self, idade: int, nome: str, altura: float):
        self.idade = idade
        self.nome = nome
        self.altura = altura

    def print_info(self):
        print(f"Idade: {self.idade}, Nome: {self.nome}, Altura: {self.altura}")
```

```python
class Cadastro(DadosPessoais):

    def __init__(self, idade: int, nome: str, altura: float, nome_da_mae: str):
        super().__init__(idade, nome, altura)
        self.nome_da_mae = nome_da_mae

    def print_info(self):
        super().print_info()
        print("Nome da mãe:", self.nome_da_mae)

# "Antes de fazer as coisas específicas de Cadastro, execute também o __init__ da classe pai."
```

```python
pessoa = Cadastro(
    idade=25,
    nome="João",
    altura=1.75,
    nome_da_mae="Maria"
)

pessoa.print_info()
```
